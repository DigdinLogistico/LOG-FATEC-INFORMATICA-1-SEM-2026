```python
from ucimlrepo import fetch_ucirepo

# fetch dataset
wine_quality = fetch_ucirepo(id=186)

# data (as pandas dataframes)
X = wine_quality.data.features
y = wine_quality.data.targets

# metadata
print(wine_quality.metadata)

# variable information
print(wine_quality.variables)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Combine the features and target into a single DataFrame for plotting
df_plot = X[['alcohol', 'fixed_acidity']].copy()
df_plot['quality'] = y

display(df_plot.head())
```

### Linear Regression Plot: Alcohol vs. Quality

```python
plt.figure(figsize=(10, 6))
sns.regplot(x='alcohol', y='quality', data=df_plot, scatter_kws={'alpha':0.3})
plt.title('Linear Regression of Wine Quality vs. Alcohol Content')
plt.xlabel('Alcohol Content')
plt.ylabel('Quality (score)')
plt.grid(True)
plt.show()
```

### Linear Regression Plot: Fixed Acidity vs. Quality

```python
plt.figure(figsize=(10, 6))
sns.regplot(x='fixed_acidity', y='quality', data=df_plot, scatter_kws={'alpha':0.3})
plt.title('Linear Regression of Wine Quality vs. Fixed Acidity')
plt.xlabel('Fixed Acidity')
plt.ylabel('Quality (score)')
plt.grid(True)
plt.show()
```
